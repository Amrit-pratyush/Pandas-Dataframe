4. Indexing using .loc

Using .loc, perform the following:


1.   Display the row with index 10.
2.   Display the Sales value of the row with index 20





```python
import pandas as pd

# Load the dataset
df = pd.read_csv('Retail_dataset.csv')

# 1. Display the row with index 10
print("--- ROW WITH INDEX 10 ---")
print(df.loc[10])

# 2. Display the Sales value of the row with index 20
sales_at_20 = df.loc[20, 'Sales']
print("\n--- SALES VALUE AT INDEX 20 ---")
print(f"Sales: {sales_at_20}")
```

5. Indexing using .iloc

Using .iloc, display:

1. The first 6 rows and first 4 columns.
2. *The* Sales value from the 3rd row.

```python
import pandas as pd

# Load the dataset
df = pd.read_csv('Retail_dataset.csv')

# 1. Display the first 6 rows (indices 0 to 5) and first 4 columns (indices 0 to 3)
print("--- FIRST 6 ROWS AND FIRST 4 COLUMNS ---")
print(df.iloc[0:6, 0:4])

# 2. Display the Sales value from the 3rd row (index 2)
# 'Sales' is at column index 4
sales_3rd_row = df.iloc[2, 4]
print("\n--- SALES VALUE FROM THE 3RD ROW ---")
print(f"Sales: {sales_3rd_row}")
```

Q7. Filtering Data

Filter the dataset to display:
1. Rows where Region = "East".
2. Rows where Sales > 1500.

```python
import pandas as pd

# Load the dataset
df = pd.read_csv('Retail_dataset.csv')

# 1. Filter rows where Region is "East"
east_region_df = df[df['Region'] == 'East']
print("--- ROWS WHERE REGION IS EAST (FIRST 5 ROWS) ---")
print(east_region_df.head())

# 2. Filter rows where Sales is greater than 1500
high_sales_df = df[df['Sales'] > 1500]
print("\n--- ROWS WHERE SALES > 1500 (FIRST 5 ROWS) ---")
print(high_sales_df.head())
```

Q8. Multiple Condition Filtering

Display rows where:
1. Region is West
2. Sales is greater than 1200

```python
import pandas as pd

# Load the dataset
df = pd.read_csv('Retail_dataset.csv')

# Apply multiple conditions using the & operator
filtered_df = df[(df['Region'] == 'West') & (df['Sales'] > 1200)]

print("--- ROWS WHERE REGION IS WEST AND SALES > 1200 (FIRST 5 ROWS) ---")
print(filtered_df.head())
```

9. Descriptive Statistics

Generate summary statistics using:

describe()

From the result identify:
1. Mean sales value
2. Maximum quantity sold
3. Minimum sales value

```python
import pandas as pd

# Load the dataset
df = pd.read_csv('Retail_dataset.csv')

# 1. Generate summary statistics for the numerical columns
summary_stats = df.describe()
print("--- SUMMARY STATISTICS ---")
print(summary_stats)

# 2. Extract specific values from the describe dataframe or directly using methods
mean_sales = df['Sales'].mean()
max_quantity = df['Quantity'].max()
min_sales = df['Sales'].min()

print("\n--- IDENTIFIED VALUES ---")
print(f"Mean Sales Value     : {mean_sales:.2f}")
print(f"Maximum Quantity Sold: {max_quantity}")
print(f"Minimum Sales Value  : {min_sales}")
```

10. Aggregation Analysis

Using the Sales column, calculate:
1. Total sales
2. Average sales
3. Maximum sales
4. Minimum sales

```python
import pandas as pd

# Load the dataset
df = pd.read_csv('Retail_dataset.csv')

# Perform aggregation tasks using built-in Pandas methods on the Sales column
total_sales = df['Sales'].sum()
average_sales = df['Sales'].mean()
maximum_sales = df['Sales'].max()
minimum_sales = df['Sales'].min()

print("--- SALES AGGREGATION ANALYSIS ---")
print(f"Total Sales   : ${total_sales:,.2f}")
print(f"Average Sales : ${average_sales:,.2f}")
print(f"Maximum Sales : ${maximum_sales:,.2f}")
print(f"Minimum Sales : ${minimum_sales:,.2f}")
```
